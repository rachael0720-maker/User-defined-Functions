```python
def test(integer,var=True,dict1={2:3, 4:5, 6:8}):
    if var==True:
        if integer in dict1.keys():
            return dict1[integer]
    
    if var==False:
        return var

print(test(6))
```

<b>Q2. Function to check is a key exist in a dictionary.

```python
def checkingIfIn(a, direction = True, d = {'mars': 2, 'jupiter': 1, 'saturn': 19, 'earth': 5, 'sun': 3, 'moon': 2, 'venus': 7}):
    if direction == True:
        if a in d:
            return d[a]
        else:
            return False
    else:
        if a not in d:
            return True
        else:
            return d[a]
```

```python
# Call the function so that it returns False and assign that function call to the variable c_false
c_false=checkingIfIn('a')
print(c_false)
```

```python
# Call the fucntion so that it returns True and assign it to the variable c_true
c_true=checkingIfIn('d',direction=False)
print(c_true)
```

```python
# Call the function so that the value of fruit is assigned to the variable fruit_ans
fruit_ans= checkingIfIn('venus')
print(fruit_ans)
```

```python
# Call the function using the first and third parameter so that the value 8 is assigned to the variable param_check
param_check = checkingIfIn('a',True, {'a': 8})
print(param_check)
```

<b>Q4. Calculate surface area of the cuboid.

```python
def surface_area_cuboid(length, height, width):
    return 2*length*height + 2*length*width + 2*height*width
```

```python
surface_area_cuboid(height=2, width=2, length=5)
```

```python
surface_area_cuboid(7,7,15)
```

<b>Q5. Function to Sort the input list.

```python
def Sort_a_List(input_list):
    for k in range(0,len(input_list)):
        for a in range(0,len(input_list)-1):
            if input_list[k]>input_list[a]:
                continue
            else:
                b=input_list[k]
                input_list[k]=input_list[a]
                input_list[a]=b
    return input_list

# If you just need the function, then skip copying the lines below!
input_list = input("Enter the list separated by space: ").split()
input_list = list(map(int,input_list))
print("Your list in Ascending order:",Sort_a_List(input_list))
```

<b> Q6. Quick sort.

```python
def quicksort(lst):
    if not lst:
        return []
    return (quicksort([x for x in lst[1:] if x <  lst[0]])
            + [lst[0]] +
            quicksort([x for x in lst[1:] if x >= lst[0]]))
```

```python
quicksort([24,71,13,42,89,95,36])
```
